```python
# CODTECH INTERNSHIP - TASK 4: TEXT GENERATION USING GPT (DistilGPT2)

from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline, set_seed
import torch

# Path where your manually downloaded model is saved
local_model_path = r"C:\Users\MUBEEN\distilgpt2_local"  # Make sure this path is correct

# Check for GPU
device = 0 if torch.cuda.is_available() else -1

# Load local model and tokenizer
print(" Loading model from local directory...")
tokenizer = AutoTokenizer.from_pretrained(local_model_path)
model = AutoModelForCausalLM.from_pretrained(local_model_path)

# Create generation pipeline
generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    device=device
)

# Set seed for consistent results
set_seed(42)

# Ask user for input (multi-line input for paragraph generation)
print("\n Welcome to the AI Text Generator!")
print(" Enter your topic or starting paragraph (Press Enter twice to finish):")

# User input for a paragraph (multi-line input)
input_paragraph = ""
while True:
    try:
        line = input()  # Takes input line-by-line
        if line == "":  # End input when the user presses Enter without any text
            break
        input_paragraph += line + "\n"  # Add each line to the paragraph
    except EOFError:
        break

# Inform user what they entered
print("\nYou entered the following paragraph:\n", input_paragraph)

# Generate output using the model
print("\n Generating paragraph...\n")
outputs = generator(
    input_paragraph,
    max_length=150,
    num_return_sequences=1,
    do_sample=True,
    temperature=0.9,
    top_k=50,
    top_p=0.95,
    truncation=True,  # Added truncation to avoid warnings
    pad_token_id=tokenizer.eos_token_id  # Added pad_token_id to fix warning
)

# Print the result
print(" Generated Paragraph:\n")
print(outputs[0]["generated_text"])
```
